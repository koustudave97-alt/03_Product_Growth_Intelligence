# Streamlit Data Contract Verification

## Product & Growth Intelligence Platform

This notebook verifies the structure of the core output datasets that will power the Streamlit application.

For each dataset, we inspect:

- File availability
- Dataset shape
- Column names
- Data types
- Sample records

This verification step ensures that the Streamlit application uses the exact exported data structure from the completed analytics notebooks.

```python
from pathlib import Path
import pandas as pd

# Project paths
PROJECT_ROOT = Path.cwd().parent
OUTPUTS_DIR = PROJECT_ROOT / "outputs"

print("Project root:", PROJECT_ROOT)
print("Outputs directory:", OUTPUTS_DIR)
print("Outputs directory exists:", OUTPUTS_DIR.exists())
```

```python
CORE_FILES = [
    "06_segment_profile.csv",
    "06_segment_funnel.csv",
    "06_segment_retention.csv",
    "07_product_intelligence.csv",
    "07_category_intelligence.csv",
    "07_product_opportunities.csv",
    "growth_opportunity_engine_output.csv"
]

print("CORE STREAMLIT DATASETS")
print("=" * 70)

for file_name in CORE_FILES:
    file_path = OUTPUTS_DIR / file_name
    status = "FOUND" if file_path.exists() else "MISSING"
    print(f"{status:8} | {file_name}")
```

```python
verification_results = {}

for file_name in CORE_FILES:
    
    file_path = OUTPUTS_DIR / file_name
    df = pd.read_csv(file_path)
    
    verification_results[file_name] = {
        "shape": df.shape,
        "columns": df.columns.tolist(),
        "dtypes": df.dtypes.astype(str).to_dict(),
        "preview": df.head(3)
    }
    
    print("\n" + "=" * 90)
    print(f"FILE: {file_name}")
    print("=" * 90)
    
    print(f"\nShape: {df.shape}")
    
    print("\nColumns:")
    for i, column in enumerate(df.columns, start=1):
        print(f"{i}. {column}")
    
    print("\nData Types:")
    print(df.dtypes)
    
    print("\nFirst 3 Rows:")
    display(df.head(3))
```

```python
assert len(verification_results) == 7, "Not all datasets were verified"

for file_name, details in verification_results.items():
    assert details["shape"][0] > 0, f"{file_name} is empty"
    assert len(details["columns"]) > 0, f"{file_name} has no columns"

print("=" * 70)
print("STREAMLIT DATA CONTRACT VALIDATION: PASSED")
print("Core datasets verified:", len(verification_results))
print("=" * 70)
```
